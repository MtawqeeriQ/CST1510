## 1. The six comparison operators

Each one asks a question and answers with `True` or `False`.

`==` equal to Â· `!=` not equal to Â· `>` greater than Â· `<` less than Â· `>=` at least Â· `<=` at most

Run the next cell.

````python
value = 23.7
limit = 20

print(value == limit)     # False  - equal to?
print(value != limit)     # True   - not equal to?
print(value >  limit)     # True   - greater than?
print(value <  limit)     # False  - less than?
print(value >= limit)     # True   - greater than or equal to?
print(value <= limit)     # False  - less than or equal to?
````

## 2. `=` stores, `==` asks

This is the single most common slip this week. `=` puts a value in. `==` checks one.

They read almost the same and do completely different things.

````python
limit = 20          # gets: "limit now holds 20"
print(limit == 20)  # asks: "does limit hold 20?"  -> True
````

### Your turn

Given `limit = 20`, write the line that **asks** whether `limit` is 20 â€” not the one that changes it.

**Answer:** `print(limit == 20)`

````python
# Your code here
limit = 20
print(limit == 20)   # asks: "does limit hold 20?" -> True
````

## 3. A comparison is a value too

The result has a type of its own: `bool`. You can store it, just like anything else.

````python
is_over = value > limit

print(is_over)
print(type(is_over))
````

## 4. Combining conditions

`and` â€” both sides must be True. `or` â€” at least one side. `not` â€” flips it.

Bracket each comparison when you combine them. It costs nothing and removes all doubt.

````python
percent = 95.0

print((percent >= 90) and (percent < 100))   # True  - inside the warning band
print(not (percent < 90))                     # True
````

## Common mistakes

| Mistake | What you see | Fix |
|---|---|---|
| `value = 20` when you meant to ask | nothing â€” it silently reassigns | use `==` to ask, `=` to store |
| `"20" == 20` | `False` | text and number are never equal â€” check the type |
| forgetting brackets in a combined condition | wrong answer, no error | bracket each comparison |
| `if 90 <= percent <= 100` before you are confident | works, but confusing this early | write `percent >= 90 and percent <= 100` until you are sure |
